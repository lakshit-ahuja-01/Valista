```python
# Curved Scores — Map, Filter Pipeline
inp = input().strip().split()
marks = list(map(int,inp))
if any(mark not in range(0,101) for mark in marks):
    print("Invalid Input")
else:
    curved = map(lambda c: min(c+5, 100), marks)
    passed = list(filter(lambda p: p >= 40, curved))
    print(passed)
```

```python
# Prime Filter — Keep Only Primes
import math

def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(math.sqrt(n))+1):
        if n%i == 0:
                return False
    return True
    
inp = list(map(int, input().split()))
if any(num < 0 for num in inp):
    print("Invalid Input")
else:
    print(list(filter(is_prime, inp)))
```

```python
# Product of Non-Zero Digits — Reduce
from functools import reduce
s = input().strip()
digits = list(map(int, s))
non_zero = list(filter(lambda d: d != 0, digits))
if len(digits)> 0 and all(d == 0 for d in digits):
    print(0)
else:
    prod = reduce(lambda a, b: a*b, non_zero, 1)
    print(prod)
```

```python
# Count Vowels and Consonants in a Sentence
s = input("Enter sentence: ").lower()
vowels = "aeiou"
v = sum(1 for ch in s if ch in vowels)
c = sum(1 for ch in s if ch.isalpha() and ch not in vowels)
print(f"Vowels: {v}")
print(f"Consonants: {c}")
```

```python
# Palindrome Checker (Ignoring Spaces & Case)
text = input().strip().lower()
if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Tuple Packing and Unpacking
value = input().split()
if len(value) != 3:
    print("Invlaid data")
else:
    name, age, city = value
    if not age.isdigit() or name.strip()=="" or city.strip()=="":
        print("invalid data")
    else:
        tup = (name, int(age), city)
        n,a,c = tup
        print(f"Name: {n}, age: {a}, city: {c}")
```
